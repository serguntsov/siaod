### 1. Найти часто встречаемый символ в строке. 
Если несколько символов встречаются одинаковое количество раз, то можно вывести любой.

Решение 1: O(N^2)

```python
s = 'd2d4537ddd98gdn43ddkgddjk4ndddddkgk' #15d
kb = 0
sb = ''

for i in range(len(s)):
    k = 0
    for j in range(len(s)):
        if s[i] == s[j]:
            k += 1
    if k > kb:
        kb = k
        sb = s[i]

print(kb, sb)
```

Решение 2: O(N*K)

```python
s = 'd2d4537ddd98gdn43ddkgddjk4ndddddkgk' #15d
kb = 0
sb = ''

for sym in set(s):
    k = 0
    for i in s:
        if sym == i:
            k += 1
    if k > kb:
        kb = k
        sb = sym

print(kb, sb)
```

Решение 3: O(2N) = O(N) 

```python
s = 'd2d4537ddd98gdn43ddkgddjk4ndddddkgk' #15d
d = {}
kb = 0
sb = ''

for sym in s:
    if sym not in d:
        d[sym] = 0
    d[sym] += 1

for k, v in d.items():
    if v > kb:
        kb = v
        sb = k

print(kb, sb)
```

Решение 4: O(N)

```python
s = 'd2d4537ddd98gdn43ddkgddjk4ndddddkgk' #15d
d = {}
kb = 0
sb = ''

for sym in s:
    if sym not in d:
        d[sym] = 0
    d[sym] += 1

    if d[sym] > kb:
        kb = d[sym]
        sb = sym
    
print(kb, sb)
```
